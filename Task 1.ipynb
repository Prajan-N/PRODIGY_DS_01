```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
import pandas as pd
df = pd.read_excel(
    '/content/API_SP.POP.TOTL_DS2_en_excel_v2_33073 (2).xls',
    engine='xlrd',
    skiprows=3
)
df.head()
```

```python
df_final = df.drop(['Country Code',	'Indicator Name',	'Indicator Code'], axis=1)
```

```python
chosen_country = 'United States'


country_data = df_final[df_final['Country Name'] == chosen_country]


years = country_data.columns[2:].astype(int)
population = country_data.iloc[:, 2:].values.flatten()

plt.figure(figsize=(5, 3))
plt.bar(years, population, color='DarkRed')
plt.xlabel('Year')
plt.ylabel('Population')
plt.title(f'Population of {chosen_country} Over the Years')


plt.show()
```

```python
plt.hist(population, bins=20, color='darkred', edgecolor='black')
plt.show()
```
